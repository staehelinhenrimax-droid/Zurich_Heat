## holoviews

```python
import geopandas as gpd
import rasterio
from rasterio.plot import reshape_as_image
import numpy as np
import holoviews as hv

hv.extension("bokeh")

# -----------------------------
# Load vector data
# -----------------------------
vector_path = "../data/raw/Stadtkreise_Zurich.gpkg"
gdf = gpd.read_file(vector_path).to_crs(epsg=32632)

# -----------------------------
# Load raster data
# -----------------------------
raster_path = "../data/raw/LandsatComposite_Zurich_2024.tif"

with rasterio.open(raster_path) as src:
    raster = src.read()
    bounds = src.bounds
    raster_crs = src.crs

# -----------------------------
# Reproject vector to raster CRS if needed
# -----------------------------
if gdf.crs != raster_crs:
    gdf = gdf.to_crs(raster_crs)

# -----------------------------
# Convert raster to image shape
# -----------------------------
img = reshape_as_image(raster)

# Keep only 3 bands for RGB display
if img.shape[2] > 3:
    img_rgb = img[:, :, [2, 3, 5]]
else:
    img_rgb = img

# -----------------------------
# Build coordinate arrays
# FIX: y must run from top to bottom
# -----------------------------
x = np.linspace(bounds.left, bounds.right, img.shape[1])
y = np.linspace(bounds.top, bounds.bottom, img.shape[0])

# -----------------------------
# Create HoloViews layers
# -----------------------------
raster_hv = hv.RGB(
    (x, y, img_rgb),
    kdims=["x", "y"]
).opts(
    width=700,
    height=500
)

vector_hv = hv.Polygons(gdf).opts(
    line_color="yellow",
    fill_alpha=0,
    line_width=2
)

# -----------------------------
# Overlay
# -----------------------------
raster_hv * vector_hv
```

## geoviews

```python
import geopandas as gpd
import rasterio
from rasterio.plot import reshape_as_image
import geoviews as gv
import cartopy.crs as ccrs

gv.extension("bokeh")

# -----------------------------
# Define CRS for GeoViews
# EPSG:32632 = UTM zone 32N
# -----------------------------
utm32 = ccrs.UTM(zone=32)

# -----------------------------
# Load vector data
# -----------------------------
vector_path = "../data/raw/Stadtkreise_Zurich.gpkg"

# Choose the polygon layer explicitly
gdf = gpd.read_file(vector_path, layer="stzh.adm_stadtkreise_v").to_crs(epsg=32632)

# -----------------------------
# Load raster data
# -----------------------------
raster_path = "../data/raw/LandsatComposite_Zurich_2024.tif"

with rasterio.open(raster_path) as src:
    raster = src.read()
    bounds = src.bounds
    raster_crs = src.crs

# -----------------------------
# Reproject vector to raster CRS if needed
# -----------------------------
if gdf.crs != raster_crs:
    gdf = gdf.to_crs(raster_crs)

# -----------------------------
# Convert raster to image shape
# -----------------------------
img = reshape_as_image(raster)

# Keep only 3 bands for RGB display
if img.shape[2] > 3:
    img_rgb = img[:, :, [2, 3, 5]]
else:
    img_rgb = img

# -----------------------------
# Create GeoViews layers
# Use bounds instead of x/y arrays
# -----------------------------
raster_gv = gv.RGB(
    img_rgb,
    bounds=(bounds.left, bounds.bottom, bounds.right, bounds.top),
    crs=utm32
).opts(
    width=700,
    height=500
)

vector_gv = gv.Polygons(
    gdf,
    crs=utm32
).opts(
    line_color="yellow",
    fill_alpha=0,
    line_width=2
)

# -----------------------------
# Overlay
# -----------------------------
raster_gv * vector_gv
```
